# Task 1: Exploring and Visualizing the Iris Dataset

## Objective
Learn how to load, inspect, and visualize a dataset to understand data trends and distributions using the Iris Dataset.

## Dataset
The Iris Dataset contains measurements for 150 iris flowers from three different species.

## Instructions
- Load the dataset using pandas.
- Print the shape, column names, and the first few rows.
- Use `.info()` and `.describe()` for summary statistics.
- Visualize the dataset using scatter plots, histograms, and box plots.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the dataset
iris = sns.load_dataset('iris')

print("Dataset Loaded Successfully!")
```

## 1. Data Inspection
We will now check the shape, column names, and the first few rows of the dataset.

```python
print("Shape of the dataset:", iris.shape)
print("\nColumn Names:", iris.columns.tolist())
print("\nFirst 5 rows:")
display(iris.head())
```

## 2. Summary Statistics
Using `.info()` and `.describe()` to understand the data types and statistical distribution.

```python
print("Dataset Info:")
iris.info()

print("\nSummary Statistics:")
display(iris.describe())
```

## 3. Data Visualization

### A. Scatter Plot
Showing relationships between features (Sepal Length vs Sepal Width).

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=iris, x='sepal_length', y='sepal_width', hue='species', style='species')
plt.title('Sepal Length vs Sepal Width')
plt.show()
```

### B. Histograms
Showing the distribution of values for each feature.

```python
iris.hist(figsize=(12, 8), bins=20)
plt.suptitle('Histograms of Iris Features')
plt.show()
```

### C. Box Plots
Identifying outliers and distribution across species.

```python
plt.figure(figsize=(12, 8))
plt.subplot(2, 2, 1)
sns.boxplot(x='species', y='sepal_length', data=iris)
plt.subplot(2, 2, 2)
sns.boxplot(x='species', y='sepal_width', data=iris)
plt.subplot(2, 2, 3)
sns.boxplot(x='species', y='petal_length', data=iris)
plt.subplot(2, 2, 4)
sns.boxplot(x='species', y='petal_width', data=iris)
plt.tight_layout()
plt.show()
```

## Conclusion
The Iris dataset is well-balanced and shows clear distinctions between species, especially when looking at petal measurements. Box plots revealed a few outliers in sepal width for the 'virginica' species.
